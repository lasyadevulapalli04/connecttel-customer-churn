# ConnectTel Telecom Customer Churn Prediction
## Notebook 02: Feature Engineering & Preprocessing Pipeline

**Objective:**
- Engineer business-specific domain features (`TotalChargesPerTenure`, `ServiceCount`, `IsMonthlyContract`, `HasTechSupport`).
- Separate target `Churn` and drop identifier `customerID` to prevent data leakage.
- Stratified 80/20 train/test dataset split.
- Construct scikit-learn `ColumnTransformer` preprocessing pipeline for scaling numeric columns and one-hot encoding categorical variables.

```python
import sys
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

sys.path.append(os.path.abspath('..'))

from src.data_preprocessing import load_raw_data, clean_telco_data, get_feature_columns, build_preprocessing_pipeline
from src.feature_engineering import create_features

print("[INFO] Environment setup complete.")
```

--- 
## 1. Load Cleaned Dataset

```python
raw_path = '../data/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df_raw = load_raw_data(raw_path)
df_clean = clean_telco_data(df_raw)
df_clean.head()
```

--- 
## 2. Apply Domain Feature Engineering

```python
df_feat = create_features(df_clean)
print("Engineered Feature Columns:", df_feat.columns.tolist())
df_feat[['tenure', 'MonthlyCharges', 'TotalCharges', 'TotalChargesPerTenure', 'ServiceCount', 'IsMonthlyContract']].head()
```

--- 
## 3. Stratified Train / Test Split

```python
X = df_feat.drop(columns=['Churn', 'customerID'], errors='ignore')
y = df_feat['Churn']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape : {X_test.shape}, y_test shape : {y_test.shape}")
print(f"Train Churn Ratio: {y_train.mean():.4f}")
print(f"Test Churn Ratio : {y_test.mean():.4f}")
```

--- 
## 4. Build Preprocessing Pipeline

```python
num_cols, cat_cols = get_feature_columns(X, target_col='', id_col='')
print("Numerical Columns  :", num_cols)
print("Categorical Columns:", cat_cols)

preprocessor = build_preprocessing_pipeline(num_cols, cat_cols)
print("[INFO] Scikit-learn ColumnTransformer preprocessor initialized.")
```

--- 
## 5. Validate Preprocessor Transformation

```python
X_train_prep = preprocessor.fit_transform(X_train)
X_test_prep = preprocessor.transform(X_test)

print("Preprocessed X_train matrix shape:", X_train_prep.shape)
print("Preprocessed X_test matrix shape :", X_test_prep.shape)
```
