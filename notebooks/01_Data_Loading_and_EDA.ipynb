# ConnectTel Telecom Customer Churn Prediction
## Notebook 01: Data Loading & Exploratory Data Analysis (EDA)

**Objective:**
- Load raw Telco customer churn dataset.
- Inspect data types, missing values, and data quality issues.
- Perform cleaning (space character conversion, missing value imputation, target binary encoding).
- Conduct univariate and bivariate Exploratory Data Analysis (EDA) to understand churn key drivers.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set project root path
sys.path.append(os.path.abspath('..'))

from src.data_preprocessing import load_raw_data, clean_telco_data

%matplotlib inline
sns.set_theme(style="whitegrid")
print("[INFO] Environment setup complete.")
```

--- 
## 1. Load Raw Dataset

```python
data_path = '../data/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df_raw = load_raw_data(data_path)
print("Dataset Shape:", df_raw.shape)
df_raw.head()
```

--- 
## 2. Inspect Data Types & Missing Values

```python
df_raw.info()
```

```python
print("Explicit Missing Values Count:")
print(df_raw.isnull().sum())

# Check for blank space strings in object columns
print("\nBlank Space String Count in TotalCharges:")
print((df_raw['TotalCharges'] == ' ').sum())
```

--- 
## 3. Clean Dataset & Encode Target Variable

```python
df_clean = clean_telco_data(df_raw)
df_clean.info()
```

--- 
## 4. Target Variable Analysis (Churn Breakdown)

```python
churn_counts = df_clean['Churn'].value_counts()
churn_pct = df_clean['Churn'].value_counts(normalize=True) * 100

print("Churn Value Counts:\n", churn_counts)
print("\nChurn Percentage:\n", churn_pct)

plt.figure(figsize=(6, 4))
sns.countplot(data=df_clean, x='Churn', palette='Set2')
plt.title('Target Variable Breakdown: Churn (0 = Retained, 1 = Churned)', fontsize=13)
plt.xlabel('Churn Status')
plt.ylabel('Customer Count')
plt.show()
```

--- 
## 5. Numerical Feature Distributions

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df_clean['tenure'], kde=True, ax=axes[0], color='royalblue', bins=30)
axes[0].set_title('Tenure Distribution (Months)')

sns.histplot(df_clean['MonthlyCharges'], kde=True, ax=axes[1], color='forestgreen', bins=30)
axes[1].set_title('Monthly Charges Distribution ($)')

sns.histplot(df_clean['TotalCharges'], kde=True, ax=axes[2], color='darkpurple' if 'darkpurple' in plt.colormaps() else 'purple', bins=30)
axes[2].set_title('Total Charges Distribution ($)')

plt.tight_layout()
plt.show()
```

--- 
## 6. Categorical Feature Relationships with Churn

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

sns.countplot(data=df_clean, x='Contract', hue='Churn', ax=axes[0, 0], palette='Blues')
axes[0, 0].set_title('Contract Type vs Churn', fontsize=12)

sns.countplot(data=df_clean, x='InternetService', hue='Churn', ax=axes[0, 1], palette='Oranges')
axes[0, 1].set_title('Internet Service Provider vs Churn', fontsize=12)

sns.countplot(data=df_clean, x='TechSupport', hue='Churn', ax=axes[1, 0], palette='Greens')
axes[1, 0].set_title('Tech Support vs Churn', fontsize=12)

sns.countplot(data=df_clean, x='PaymentMethod', hue='Churn', ax=axes[1, 1], palette='Reds')
axes[1, 1].set_title('Payment Method vs Churn', fontsize=12)
axes[1, 1].tick_params(axis='x', rotation=25)

plt.tight_layout()
plt.show()
```

## 7. EDA Findings & Summary Observations

*Execute the notebook cells above manually and note your observations here:*
- **Target Class Ratio:** ~26.5% Churn vs 73.5% Retained (Mild class imbalance).
- **Contract Impact:** Month-to-month contract customers exhibit significantly higher churn risk than 1-Year or 2-Year contract holders.
